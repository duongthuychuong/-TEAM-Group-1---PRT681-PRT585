```python
# Import datetime for date handling
from datetime import datetime
```

```python
# ========== CLASS 1: Base Class (Parent) ==========
# This shows INHERITANCE and ENCAPSULATION

class Report:
    # Constructor
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self._created_date = None  # Private variable (ENCAPSULATION)
        self.set_created_date()    # Set creation date
    
    # Method to set date - shows ENCAPSULATION
    def set_created_date(self):
        self._created_date = datetime.now()
    
    # Getter for created date (read-only)
    def get_created_date(self):
        return self._created_date
    
    # Virtual method - can be overridden by child classes (POLYMORPHISM)
    def generate(self):
        print(f"Generating report: {self.title}")
        print(f"Author: {self.author}")
        print(f"Date: {self.get_created_date()}")
    
    # Method to display report info
    def display_info(self):
        print("=" * 50)
        print(f"REPORT: {self.title}")
        print(f"Author: {self.author}")
        if self.get_created_date():
            print(f"Created: {self.get_created_date().strftime('%Y-%m-%d %H:%M')}")
        print("=" * 50)
```

```python
# ========== CLASS 2: Child Class (Sales Report) ==========
# This shows INHERITANCE - SalesReport INHERITS from Report

class SalesReport(Report):
    def __init__(self, title, author, total_sales, number_of_orders):
        super().__init__(title, author)  # Call parent constructor
        self.total_sales = total_sales
        self.number_of_orders = number_of_orders
    
    # Override the generate method - shows POLYMORPHISM
    def generate(self):
        super().generate()  # Call parent method
        print(f"Total Sales: ${self.total_sales:,.2f}")
        print(f"Number of Orders: {self.number_of_orders}")
        if self.number_of_orders > 0:
            avg = self.total_sales / self.number_of_orders
            print(f"Average Order Value: ${avg:,.2f}")
```

```python
# ========== CLASS 3: Another Child Class (Inventory Report) ==========

class InventoryReport(Report):
    def __init__(self, title, author, total_items, low_stock_items):
        super().__init__(title, author)
        self.total_items = total_items
        self.low_stock_items = low_stock_items
    
    def generate(self):
        super().generate()
        print(f"Total Items: {self.total_items}")
        print(f"Low Stock Items: {self.low_stock_items}")
        if self.total_items > 0:
            health = ((self.total_items - self.low_stock_items) / self.total_items) * 100
            print(f"Stock Health: {health:.0f}%")
```

```python
# ========== MAIN PROGRAM ==========

print("=" * 60)
print("     REPORT GENERATOR DEMO - OOP Concepts")
print("=" * 60)
print("Demonstrating: Inheritance, Polymorphism, Encapsulation\n")

# Create a list of reports - shows POLYMORPHISM
reports = []

# Adding different types of reports
reports.append(SalesReport("Q1 2026 Sales", "Bibek Prasai", 125000.50, 450))
reports.append(InventoryReport("Warehouse Stock", "Bibek Prasai", 5000, 120))
reports.append(SalesReport("Q2 2026 Sales", "Bibek Prasai", 142000.75, 520))

# Generate all reports - POLYMORPHISM in action!
for report in reports:
    report.display_info()
    report.generate()
    print("-" * 50)
    print()
```

```python
# ========== ENCAPSULATION DEMO ==========

print("\n" + "=" * 60)
print("     ENCAPSULATION DEMO")
print("=" * 60)

my_report = Report("My Report", "Bibek")
my_report.title = "Updated Report Title"  # Can change title (public)
# my_report._created_date = "2026-01-01"  # Cannot change easily (private)

print(f"Report Title: {my_report.title}")
print(f"Created Date: {my_report.get_created_date()} (Read-only)")
print("\n✅ Encapsulation: The created date is private and cannot be changed directly.")
```
